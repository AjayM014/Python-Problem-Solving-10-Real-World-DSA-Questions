```python
students = list(map(int, input("Enter student IDs: ").split()))

seen = set()
left = 0
maximum = 0

for right in range(len(students)):
    while students[right] in seen:
        seen.remove(students[left])
        left += 1

    seen.add(students[right])
    maximum = max(maximum, right - left + 1)

print("Maximum length:", maximum)
```

```python
discounts = list(map(int, input("Enter discount values: ").split()))

current = discounts[0]
maximum = discounts[0]

for i in range(1, len(discounts)):
    current = max(discounts[i], current + discounts[i])
    maximum = max(maximum, current)

print("Maximum discount value:", maximum)
```

```python
heights = list(map(int, input("Enter building heights: ").split()))

left = 0
right = len(heights) - 1

left_max = 0
right_max = 0

water = 0

while left < right:
    if heights[left] <= heights[right]:

        if heights[left] >= left_max:
            left_max = heights[left]
        else:
            water += left_max - heights[left]

        left += 1

    else:

        if heights[right] >= right_max:
            right_max = heights[right]
        else:
            water += right_max - heights[right]

        right -= 1

print("Total trapped water:", water)
```

```python
scores = list(map(int, input("Enter performance scores: ").split()))

current = scores[0]
maximum = scores[0]

for i in range(1, len(scores)):
    current = max(scores[i], current + scores[i])
    maximum = max(maximum, current)

print("Highest overall performance:", maximum)
```

```python
sales = list(map(int, input("Enter sales values: ").split()))

maximum = sales[0]
minimum = sales[0]
answer = sales[0]

for i in range(1, len(sales)):

    if sales[i] < 0:
        maximum, minimum = minimum, maximum

    maximum = max(sales[i], maximum * sales[i])
    minimum = min(sales[i], minimum * sales[i])

    answer = max(answer, maximum)

print("Maximum product:", answer)
```

```python
products = list(map(int, input("Enter product IDs: ").split()))

seen = set()
left = 0
maximum = 0

for right in range(len(products)):

    while products[right] in seen:
        seen.remove(products[left])
        left += 1

    seen.add(products[right])

    length = right - left + 1

    if length > maximum:
        maximum = length

print("Longest unique purchase sequence:", maximum)
```

```python
transactions = list(map(int, input("Enter transactions: ").split()))
target = int(input("Enter target amount: "))

prefix_sum = 0
count = 0

frequency = {0: 1}

for amount in transactions:

    prefix_sum += amount

    required = prefix_sum - target

    if required in frequency:
        count += frequency[required]

    frequency[prefix_sum] = frequency.get(prefix_sum, 0) + 1

print("Number of continuous groups:", count)
```

```python
words = input("Enter skill codes: ").split()

groups = {}

for word in words:

    key = ''.join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print("Skill groups:")

for group in groups.values():
    print(group)
```

```python
packets = list(map(int, input("Enter packet identifiers: ").split()))

packet_set = set(packets)

maximum = 0

for packet in packet_set:

    if packet - 1 not in packet_set:

        current = packet
        length = 1

        while current + 1 in packet_set:
            current += 1
            length += 1

        maximum = max(maximum, length)

print("Longest consecutive sequence:", maximum)
```

```python
n = int(input("Enter number of appointments: "))

appointments = []

for i in range(n):
    start, end = map(int, input("Enter start and end time: ").split())
    appointments.append([start, end])

appointments.sort()

merged = []

for appointment in appointments:

    if not merged or appointment[0] > merged[-1][1]:
        merged.append(appointment)

    else:
        merged[-1][1] = max(merged[-1][1], appointment[1])

print("Merged appointment schedule:")

for appointment in merged:
    print(appointment[0], appointment[1])
```
